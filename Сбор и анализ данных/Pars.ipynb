```python
import pandas as pd
import numpy as np
import csv
from vk_api import VkApi
from vk_api.exceptions import ApiError
from datetime import datetime
```

```python
with open('token.txt') as file:  
    token = file.readline()  

domain = 'shastoon.channel' # выбрала вот этого пользователя https://vk.com/shastoon.channel
```

```python
vk_session = VkApi(token=token)
vk = vk_session.get_api()

offset = 0
count = 10

try:
    wall_posts = vk.wall.get(domain = domain,
                             count = count,
                             offset = offset)
    
    data_posts = []
    cnt_posts = wall_posts['count']

    while offset < cnt_posts:

        for post in wall_posts['items']:
            data_posts.append(
                {
                    'Date': datetime.fromtimestamp(post['date']), 
                    'Likes': post['likes']['count']
                }
            )

        offset += count
        wall_posts = vk.wall.get(domain = domain,
                                 count = count,
                                 offset = offset)

    for post in wall_posts['items']:
        data_posts.append(
            {
                'Date': datetime.fromtimestamp(post['date']), 
                'Likes': post['likes']['count']
            }
        )
except ApiError as e:
    print(f"Ошибка VK API: {e}")
```

```python

# Запись полученных данных в файл 'data.csv'
import csv

with open('data.csv', 'w', newline = '', encoding = 'utf-8-sig') as file:
    fieldnames = ['Date', 'Likes']
    writer = csv.DictWriter(file, fieldnames = fieldnames)

    writer.writeheader()
    writer.writerows(data_posts)
```
